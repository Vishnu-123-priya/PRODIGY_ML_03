```python
import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

DATA_DIR = "../data/PetImages"

cat_dir = os.path.join(DATA_DIR, "Cat")
dog_dir = os.path.join(DATA_DIR, "Dog")

print("Cat images:", len(os.listdir(cat_dir)))
print("Dog images:", len(os.listdir(dog_dir)))
```

```python
cat_images = os.listdir(cat_dir)
dog_images = os.listdir(dog_dir)

fig, axes = plt.subplots(2, 4, figsize=(12, 6))

for i in range(4):
    cat_path = os.path.join(cat_dir, cat_images[i])
    dog_path = os.path.join(dog_dir, dog_images[i])

    cat_img = Image.open(cat_path)
    dog_img = Image.open(dog_path)

    axes[0, i].imshow(cat_img)
    axes[0, i].set_title("Cat")
    axes[0, i].axis("off")

    axes[1, i].imshow(dog_img)
    axes[1, i].set_title("Dog")
    axes[1, i].axis("off")

plt.tight_layout()
plt.show()
```
